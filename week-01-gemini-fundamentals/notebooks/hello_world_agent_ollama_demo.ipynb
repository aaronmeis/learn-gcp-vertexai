# Hello World Agent - Ollama Demo (Truly Local)

This notebook demonstrates how to use Ollama for truly local AI agent development - no internet required after setup!

## What is Ollama?

Ollama allows you to run large language models (LLMs) completely locally on your machine. Unlike cloud-based solutions, Ollama:
- ✅ Works offline (no internet required)
- ✅ No API keys needed
- ✅ Complete privacy (data stays local)
- ✅ Free and open source


## Prerequisites

Before running this notebook, make sure:

1. **Ollama is installed**: https://ollama.ai
2. **Ollama service is running** (usually starts automatically)
3. **A model is pulled** (e.g., `ollama pull llama2`)

You can check if Ollama is running by opening a terminal and running:
```bash
ollama list
```


## Setup and Imports


````python
import os
import requests
import json
from typing import Optional

# Configuration
OLLAMA_BASE_URL = os.getenv("OLLAMA_BASE_URL", "http://localhost:11434")
OLLAMA_MODEL = os.getenv("OLLAMA_MODEL", "llama2")  # Change to your preferred model

print(f"Ollama URL: {OLLAMA_BASE_URL}")
print(f"Model: {OLLAMA_MODEL}")
````

## Check Ollama Connection


````python
def check_ollama_running() -> bool:
    """Check if Ollama is running locally."""
    try:
        response = requests.get(f"{OLLAMA_BASE_URL}/api/tags", timeout=2)
        if response.status_code == 200:
            models = response.json().get("models", [])
            print(f"[OK] Ollama is running!")
            print(f"[OK] Available models: {[m.get('name', 'unknown') for m in models]}")
            return True
        return False
    except Exception as e:
        print(f"[ERROR] Ollama is not running: {e}")
        print("\n[INFO] To start Ollama:")
        print("  1. Install from: https://ollama.ai")
        print("  2. Start the Ollama service")
        print("  3. Pull a model: ollama pull llama2")
        return False

# Check connection
if check_ollama_running():
    print("\n✅ Ready to use Ollama!")
else:
    print("\n❌ Please start Ollama before continuing")
````

## Chat with Ollama


````python
def chat_with_ollama(prompt: str, model: str = None) -> Optional[str]:
    """Send a prompt to Ollama and get a response."""
    if not model:
        model = OLLAMA_MODEL
    
    try:
        response = requests.post(
            f"{OLLAMA_BASE_URL}/api/generate",
            json={
                "model": model,
                "prompt": prompt,
                "stream": False
            },
            timeout=60  # Longer timeout for first request
        )
        
        if response.status_code == 200:
            return response.json().get("response", "")
        else:
            return f"[ERROR] Ollama API error: {response.status_code}"
    except requests.exceptions.ConnectionError:
        return None
    except Exception as e:
        return f"[ERROR] {str(e)}"

# Test the function
test_prompt = "Hello! Can you introduce yourself?"
print(f"[USER] {test_prompt}")
response = chat_with_ollama(test_prompt)
if response:
    print(f"[AGENT] {response}")
else:
    print("[ERROR] Failed to get response")
````

## Create Hello World Agent


````python
# System prompt for our Hello World agent
system_prompt = """You are a helpful Hello World agent demonstrating local AI capabilities.
You can:
- Answer questions about AI and machine learning
- Help users learn about local LLMs
- Have friendly conversations
- Explain the benefits of running AI models locally

Keep responses concise and friendly."""

print("System prompt configured:")
print("-" * 50)
print(system_prompt)
print("-" * 50)
````

## Example Interactions


````python
# Example interactions with the agent
interactions = [
    "Hello! What can you do?",
    "Tell me about running AI models locally.",
    "How is this different from cloud AI?",
    "What are the advantages of using Ollama?"
]

print("=" * 70)
print("Hello World Agent - Example Interactions")
print("=" * 70)

for i, user_input in enumerate(interactions, 1):
    print(f"\n[{i}] [USER] {user_input}")
    print("-" * 70)
    
    # Combine system prompt with user input
    full_prompt = f"{system_prompt}\n\nUser: {user_input}\nAssistant:"
    
    # Get response from Ollama
    response = chat_with_ollama(full_prompt)
    
    if response:
        print(f"[AGENT] {response}")
    else:
        print("[ERROR] Failed to get response from Ollama")
    
    print("=" * 70)
````

## Interactive Chat Function

Use this function to have a conversation with the agent:


````python
def interactive_chat(user_input: str):
    """Interactive chat function for notebook use."""
    full_prompt = f"{system_prompt}\n\nUser: {user_input}\nAssistant:"
    response = chat_with_ollama(full_prompt)
    
    if response:
        print(f"[USER] {user_input}")
        print(f"[AGENT] {response}")
        return response
    else:
        print("[ERROR] Failed to get response")
        return None

# Example usage:
# interactive_chat("What is machine learning?")
````

## Try It Yourself!

Use the `interactive_chat()` function to ask your own questions:


````python
# Try your own question here!
interactive_chat("Explain the difference between local and cloud AI in one sentence.")
````
