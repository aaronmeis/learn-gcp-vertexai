# Hello World Agent - Gemini API Demo

This notebook demonstrates how to use Google's Gemini API for AI agent development - simpler than Vertex AI, just needs an API key!

## What is Gemini API?

Google's Gemini API provides direct access to Gemini models with a simple API key. It's:
- ✅ Easy to set up (just an API key)
- ✅ No GCP project needed
- ✅ Free tier available
- ✅ Similar simplicity to Ollama, but cloud-based


## Prerequisites

1. **Get a free Gemini API key**: https://makersuite.google.com/app/apikey
2. **Add to `.env` file**: `GEMINI_API_KEY=your-api-key-here`
3. **Or set environment variable**: `export GEMINI_API_KEY=your-key`


## Setup and Configuration


```python
import os
import google.generativeai as genai
from dotenv import load_dotenv

# Load environment variables from .env file
load_dotenv()

# Configuration
API_KEY = os.getenv("GEMINI_API_KEY", "")
MODEL_NAME = os.getenv("GEMINI_MODEL", "gemini-pro")  # or "gemini-1.5-flash" for faster responses

print(f"Model: {MODEL_NAME}")
if API_KEY:
    print("[OK] API key found")
    genai.configure(api_key=API_KEY)
else:
    print("[WARNING] GEMINI_API_KEY not found!")
    print("Get your free API key from: https://makersuite.google.com/app/apikey")
    print("Then add it to your .env file: GEMINI_API_KEY=your-api-key-here")
```

## Initialize Gemini Model


```python
if API_KEY:
    # Create the model instance
    model = genai.GenerativeModel(MODEL_NAME)
    print(f"[OK] Gemini model '{MODEL_NAME}' initialized successfully!")
else:
    model = None
    print("[ERROR] Cannot initialize model without API key")
```

## Create Hello World Agent


```python
# System instruction for our Hello World agent
system_instruction = """You are a helpful Hello World agent demonstrating Google's Gemini AI capabilities.
You can:
- Answer questions about Vertex AI and Google Cloud
- Have friendly conversations
- Help users get started with AI development
- Explain the benefits of using Gemini API

Keep responses concise and friendly."""

print("System instruction configured:")
print("-" * 50)
print(system_instruction)
print("-" * 50)
```

## Chat Function


```python
def chat_with_agent(user_input: str):
    """Send a message to the agent and get a response."""
    if not model:
        return "[ERROR] Model not initialized. Please set GEMINI_API_KEY."
    
    try:
        # Combine system instruction with user input
        prompt = f"{system_instruction}\n\nUser: {user_input}\nAgent:"
        
        response = model.generate_content(prompt)
        return response.text
    except Exception as e:
        return f"[ERROR] Failed to get response: {str(e)}"

# Test the function
if model:
    test_prompt = "Hello! Can you introduce yourself?"
    print(f"[USER] {test_prompt}")
    response = chat_with_agent(test_prompt)
    print(f"[AGENT] {response}")
else:
    print("[SKIP] Model not initialized - skipping test")
```

## Example Interactions


```python
if model:
    # Example interactions
    interactions = [
        "Hello! What can you do?",
        "What is Vertex AI?",
        "How is Gemini API different from Vertex AI?",
        "What are the advantages of using Gemini API?"
    ]
    
    print("=" * 70)
    print("Hello World Agent - Example Interactions")
    print("=" * 70)
    
    for i, user_input in enumerate(interactions, 1):
        print(f"\n[{i}] [USER] {user_input}")
        print("-" * 70)
        response = chat_with_agent(user_input)
        print(f"[AGENT] {response}")
        print("=" * 70)
else:
    print("[SKIP] Model not initialized - please set GEMINI_API_KEY")
```

## Try It Yourself!

Use the `chat_with_agent()` function to ask your own questions:


```python
# Try your own question here!
if model:
    chat_with_agent("Explain the difference between Gemini API and Vertex AI in one sentence.")
else:
    print("[SKIP] Model not initialized")
```
