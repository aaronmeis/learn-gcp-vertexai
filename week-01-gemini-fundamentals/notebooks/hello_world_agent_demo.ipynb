# Hello World Agent - Vertex AI Agent Builder Demo

This notebook demonstrates how to create and interact with a Hello World agent using Vertex AI Agent Builder.


## Setup and Configuration


```python
import os
import vertexai
from vertexai.preview import agent_builder

# Configuration
PROJECT_ID = os.getenv("GOOGLE_CLOUD_PROJECT", "your-project-id")
LOCATION = os.getenv("GOOGLE_CLOUD_LOCATION", "us-central1")

print(f"Project ID: {PROJECT_ID}")
print(f"Location: {LOCATION}")
```

## Initialize Vertex AI


```python
# Initialize Vertex AI
vertexai.init(project=PROJECT_ID, location=LOCATION)
print("✓ Vertex AI initialized successfully!")
```

## Agent Configuration


```python
# Define agent configuration
agent_config = {
    "display_name": "Hello World Agent",
    "default_language_code": "en",
    "time_zone": "America/New_York",
    "description": "A simple Hello World agent demonstrating Vertex AI Agent Builder"
}

print("Agent Configuration:")
for key, value in agent_config.items():
    print(f"  {key}: {value}")
```

## Define Intents and Responses


```python
# Define intents for the Hello World agent
intents = {
    "greeting": {
        "training_phrases": ["hello", "hi", "hey", "good morning"],
        "response": "Hello! Welcome to Vertex AI Agent Builder. How can I help you today?"
    },
    "about_vertex_ai": {
        "training_phrases": ["what is vertex ai", "tell me about vertex ai"],
        "response": "Vertex AI is Google Cloud's unified ML platform for building and deploying AI agents."
    },
    "capabilities": {
        "training_phrases": ["what can you do", "what are your capabilities"],
        "response": "I'm a Hello World agent demonstrating Vertex AI capabilities!"
    }
}

print("Defined Intents:")
for intent_name, intent_data in intents.items():
    print(f"\n  {intent_name.upper()}")
    print(f"    Training Phrases: {', '.join(intent_data['training_phrases'])}")
    print(f"    Response: {intent_data['response']}")
```

## Simulate Agent Interactions


```python
# Simulate interactions with the agent
def simulate_interaction(user_input):
    """Simulate an agent interaction based on user input."""
    user_input_lower = user_input.lower()
    
    # Simple intent matching
    for intent_name, intent_data in intents.items():
        for phrase in intent_data["training_phrases"]:
            if phrase in user_input_lower:
                return intent_data["response"]
    
    return "I'm not sure how to respond to that. Can you try rephrasing?"

# Test interactions
test_inputs = [
    "Hello",
    "What is Vertex AI?",
    "What can you do?",
    "Tell me about your capabilities"
]

print("Agent Interactions:\n")
print("=" * 50)
for user_input in test_inputs:
    response = simulate_interaction(user_input)
    print(f"\n👤 User: {user_input}")
    print(f"🤖 Agent: {response}")
print("\n" + "=" * 50)
```
